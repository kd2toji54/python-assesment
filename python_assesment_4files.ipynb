## 2. Your first program: `print()`
`print()` is a **function** that shows something on the screen. Whatever you put inside the round brackets is printed.
Text must be inside quotes - double `" "` or single `' '` - so that Python knows it is text and not a command.

Run the cell below with **Shift + Enter**.

````python
# print() shows the text inside the brackets on the screen
print("Hello, Bannari Amman!")      # text in double quotes
print('Welcome to Python')          # single quotes work the same way
print("I am learning Python for AI")
````

> **What happened?** Each `print()` showed its text on a **new line**. The quotes themselves are not printed - they only mark where the text starts and ends.

`print()` can also show **numbers** and the **result of a calculation**. Numbers do not need quotes.
You can print **several values at once** by separating them with commas.

````python
print(25)                              # a number - no quotes needed
print(10 + 5)                          # Python calculates first, then prints 15
print("My age is", 18)                 # text and a number, separated by a comma
print("Total marks:", 85 + 90 + 78)    # text and a calculation
````

> **What happened?** `10 + 5` was **calculated** and only the answer `15` was printed. When values are separated by commas, `print()` puts a **space** between them automatically.

### `sep` and `end`
`print()` has two useful settings:
- `sep` (separator) decides what goes **between** values. Default: a space.
- `end` decides what goes at the **end** of the line. Default: a new line, which is why each `print()` starts on a new line.

````python
print("01", "08", "2026", sep="-")   # put "-" between the values instead of a space
print("Loading", end="...")          # end with "..." instead of going to a new line
print("done!")                       # so this continues on the same line
````

> **What happened?** The date was joined with `-` because of `sep="-"`. The word `done!` appeared on the **same line** as `Loading...` because `end="..."` stopped the first `print()` from moving to a new line.

## 3. Variables
A **variable** is a named box in the computer's memory that stores a value. We use the `=` sign to put a value into the box.

```
name = "Asha"      # create a box called name and put "Asha" in it
```
**Note:** in Python, `=` means **store** (assign), not *equal to*. Read `age = 18` as *"age gets 18"*.

````python
# Create three variables and store values in them
name = "Asha"      # a piece of text
age = 18           # a whole number
cgpa = 8.6         # a decimal number

# Use the variable names to print their values (no quotes around variable names!)
print(name)
print(age)
print(cgpa)
print(name, "is", age, "years old and has a CGPA of", cgpa)
````

> **What happened?** Python remembered the values and printed them when we used the variable names. Notice: `print(name)` printed **Asha**, but `print("name")` would print the word **name**, because quotes turn it into plain text.

A variable can be **changed** at any time. Storing a new value **replaces** the old one - the box can hold only one value at a time.

````python
attendance = 70                   # first value
print("Attendance:", attendance)

attendance = 76                   # new value replaces 70
print("Attendance now:", attendance)
````

> **What happened?** The first `print()` showed 70. After the second assignment, the box `attendance` holds 76 and the old value 70 is gone.

### Rules for variable names
| Allowed | Not allowed |
|---|---|
| `marks`, `total_marks`, `mark1`, `_count` | `1mark` (cannot start with a number) |
| letters, numbers, underscore `_` | `total marks` (no spaces), `total-marks` (no `-`) |
| | Python keywords such as `if`, `for`, `while`, `class` |

- Names are **case sensitive**: `Marks` and `marks` are two different variables.
- Use meaningful names: `hostel_fee` is much better than `x`.
- Never use `print`, `input` or `type` as variable names - it breaks those functions.

### Multiple assignment and swapping
Python lets you give values to several variables in **one line**. It also has a very easy way to **swap** two values.

````python
a, b, c = 10, 20, 30     # a gets 10, b gets 20, c gets 30 (matched by position)
print(a, b, c)

x = 5
y = 9
x, y = y, x              # swap: x gets y's value and y gets x's value at the same time
print("x =", x, " y =", y)
````

> **What happened?** Values on the right are given to the names on the left **in order**. In the swap, Python first reads the right side (`9, 5`) and then stores them, so `x` becomes 9 and `y` becomes 5. In many other languages you need a third temporary variable to do this.

## 4. Data types
Every value in Python has a **type**. The type decides what you can do with the value (for example, you can divide numbers but not names).

| Type | Meaning | Examples |
|---|---|---|
| `int` | integer - whole numbers | `18`, `-5`, `2026` |
| `float` | numbers with a decimal point | `8.6`, `3.14`, `-0.5` |
| `str` | string - text in quotes | `"Asha"`, `'CSE'`, `"123"` |
| `bool` | boolean - only two values | `True`, `False` |

The function `type()` tells you the type of any value.

````python
roll_no = 101          # whole number      -> int
cgpa = 8.6             # decimal number    -> float
dept = "AI and DS"     # text in quotes    -> str
is_hostel = True       # True or False     -> bool (no quotes, capital T)

print(type(roll_no))
print(type(cgpa))
print(type(dept))
print(type(is_hostel))
````

> **What happened?** Python shows each type as `<class '...'>`. You will often check types like this when your program gives an unexpected result.

**Careful:** a number inside quotes, like `"123"`, is a **string** (text), not a number. Watch what `+` does with text and with numbers.

````python
print(type("123"))   # in quotes -> str
print(type(123))     # no quotes -> int
print("10" + "5")    # + on two strings JOINS them   -> "105"
print(10 + 5)        # + on two numbers ADDS them    -> 15
````

> **What happened?** `"10" + "5"` gave `105` because `+` **joins** (concatenates) strings. `10 + 5` gave `15` because `+` **adds** numbers. The same symbol does different jobs depending on the type - so types matter!
> 
> There is also a special value `None`, which means *no value yet*, for example `result = None`.

## 5. Type conversion
Sometimes we must change a value from one type to another. Python gives us conversion functions:
- `int()` - convert to a whole number
- `float()` - convert to a decimal number
- `str()` - convert to text

````python
marks_text = "87"              # this is text
marks = int(marks_text)        # convert text -> int so we can do maths
print(marks + 3)               # 90

price = float("49.50")         # convert text -> float
print(price * 2)               # 99.0

roll = 101
message = "Roll number " + str(roll)   # convert int -> str so it can be joined with text
print(message)

print(int(7.9))      # float -> int simply CUTS OFF the decimal part -> 7
print(round(7.9))    # round() rounds to the NEAREST whole number    -> 8
````

> **What happened?** After conversion we could do maths with `"87"` and join `101` with text. Note the difference: `int(7.9)` gives **7** (it chops), but `round(7.9)` gives **8** (it rounds).
> 
> Not every conversion works: `int("hello")` gives a `ValueError`, because "hello" is not a number.

In Colab, a small text box appears under the cell - click it, type, and press Enter.

````python
age = int(input("Enter your age: "))    # input gives text; int() converts it to a number
print("Next year you will be", age + 1) # now we can add 1
````

## 7. Arithmetic operators
| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide (always gives a float) | `7 / 2` | `3.5` |
| `//` | floor division (only the whole part) | `7 // 2` | `3` |
| `%` | modulus (the remainder) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

Run the cell and compare each result with the table.

````python
a = 17
b = 5
print("a + b  =", a + b)    # 22
print("a - b  =", a - b)    # 12
print("a * b  =", a * b)    # 85
print("a / b  =", a / b)    # 3.4  (normal division, gives a decimal)
print("a // b =", a // b)   # 3    (5 goes into 17 three whole times)
print("a % b  =", a % b)    # 2    (remainder: 17 - 3*5 = 2)
print("a ** b =", a ** b)   # 1419857 (17 to the power 5)
````

> **What happened?** `/` gives the exact answer `3.4`. `//` gives only the **whole part** `3`, and `%` gives what is **left over** `2`. Think of 17 chocolates shared among 5 friends: each gets `17 // 5 = 3` and `17 % 5 = 2` are left.

### Order of operations
Python follows the same rule you learned in maths (**BODMAS**): brackets first, then `**`, then `* / // %`, then `+ -`.
When in doubt, **use brackets** - they make your meaning clear.

````python
print(2 + 3 * 4)       # multiplication first: 2 + 12 = 14
print((2 + 3) * 4)     # brackets first: 5 * 4 = 20

# Average of three marks - brackets are essential!
m1, m2, m3 = 80, 90, 70
print("Wrong average:", m1 + m2 + m3 / 3)     # only m3 is divided by 3
print("Right average:", (m1 + m2 + m3) / 3)   # the whole total is divided by 3
````

> **What happened?** Without brackets Python calculated `80 + 90 + (70 / 3)` = 193.33, which is wrong. With brackets it first added all marks (240) and then divided by 3 = 80.0. A very common beginner mistake!

### Real example: splitting the mess bill
Four friends share a bill of Rs. 1450. How much does each pay? And how would the bill be paid with Rs. 100 notes?

````python
total_bill = 1450      # rupees
friends = 4

share = total_bill / friends          # normal division -> 362.5
print("Each person pays Rs.", share)

notes = total_bill // 100             # how many full 100-rupee notes -> 14
change = total_bill % 100             # what is left over             -> 50
print(notes, "notes of Rs.100 and Rs.", change, "in change")
````

> **What happened?** `/` gave the exact share. `//` and `%` together split 1450 into 14 hundreds and 50 rupees - the same idea is used to convert minutes into hours, or seconds into minutes.

### `%` is very useful
Two tricks you will use again and again:
- `n % 2 == 0` is True when `n` is **even** (no remainder when divided by 2).
- `n % 10` gives the **last digit** of `n`.

````python
n = 2026
print("Last digit:", n % 10)        # 2026 divided by 10 leaves remainder 6
print("Is even?", n % 2 == 0)       # remainder 0 when divided by 2 -> True
````

> **What happened?** 2026 ÷ 10 = 202 remainder **6**, so the last digit is 6. 2026 ÷ 2 leaves remainder 0, so it is even and the comparison gives `True`.

## 8. Assignment shortcuts
We often update a variable using its own old value. Python has shortcuts:

| Shortcut | Same as |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |

````python
steps = 0
steps += 1000      # walked to the canteen:  steps = 0 + 1000
steps += 2500      # walked to the library:  steps = 1000 + 2500
print("Steps today:", steps)

balance = 500
balance -= 120     # spent Rs. 120 on lunch: balance = 500 - 120
print("UPI balance:", balance)
````

> **What happened?** Each `+=` added to the old value, so steps became 3500. `-=` reduced the balance to 380. These shortcuts are used a lot inside loops (Notebook 2).

## 9. Comparison operators
A comparison asks a **question** whose answer is always `True` or `False` (a `bool`).

| Operator | Question |
|---|---|
| `==` | Are they equal? (**two** `=` signs) |
| `!=` | Are they different? |
| `>` `<` | Greater than? Less than? |
| `>=` `<=` | Greater than or equal? Less than or equal? |

````python
attendance = 78
print(attendance >= 75)    # is 78 at least 75?   -> True
print(attendance == 100)   # is it exactly 100?   -> False
print(attendance != 0)     # is it different from 0? -> True
print("cat" == "Cat")      # text comparison is case sensitive -> False
````

> **What happened?** Each line printed the answer to a yes/no question. Remember: `=` **stores** a value, `==` **compares** two values. Mixing them up is the most common beginner error.

## 10. Logical operators: `and`, `or`, `not`
They combine True/False answers:
- `and` - True only if **both** sides are True
- `or` - True if **at least one** side is True
- `not` - reverses the answer (True becomes False, False becomes True)

Example: a student can write the exam only if attendance is at least 75 **and** internal marks are at least 40.

````python
attendance = 80
internal_marks = 42

eligible = attendance >= 75 and internal_marks >= 40   # both conditions must be True
print("Eligible for exam:", eligible)

is_weekend = False
is_holiday = True
print("No class today:", is_weekend or is_holiday)        # one True is enough -> True
print("Class today:", not (is_weekend or is_holiday))     # not reverses it      -> False
````

> **What happened?** Both conditions were True, so `eligible` is True. Change `internal_marks` to 30 and run again - it becomes False because `and` needs **both** sides to be True.

### AI connection
In Week 1 you saw a spam filter. A very simple **rule-based** spam check could combine clues with `and`:

````python
# Three clues about an email (True = the clue is present)
has_link = True
from_unknown_sender = True
says_you_won = True

# The email looks like spam only if ALL three clues are present
looks_like_spam = has_link and from_unknown_sender and says_you_won
print("Spam?", looks_like_spam)
````

> **What happened?** All three clues were True, so the result is True. Real AI does **not** use fixed rules like this - it **learns** which clues matter from thousands of labelled emails. But its final answer is still a True/False value like this one!

## 11. Neat output with f-strings
An **f-string** is the easiest way to mix text and variables. Write `f` before the opening quote and put variable names inside curly brackets `{ }`.
- `{average:.2f}` shows the value with **2 decimal places**.
- You can even write a calculation inside `{ }`.

````python
name = "Ravi"
total = 437
subjects = 5
average = total / subjects           # 87.4

print(f"{name} scored {total} marks in {subjects} subjects.")   # values go into the {}
print(f"Average = {average}")
print(f"Average (2 decimals) = {average:.2f}")                   # :.2f -> 2 decimal places
print(f"Percentage = {total / 500 * 100:.1f}%")                  # calculation inside {}
````

> **What happened?** Python replaced each `{ }` with its value. f-strings are cleaner than joining text with `+` and `str()`, and you will use them in almost every program.

## Mini project: Mess bill calculator
**Mini project.** Write a program that reads:
- number of days a student ate in the mess this month
- cost per day (Rs.)
- number of special dinners (Rs. 60 extra each)

Print a neat bill showing the **food cost**, **special dinner cost**, **5% GST** on the total, and the **grand total**, all with 2 decimal places.

<details><summary><b>Hint</b> (click to open)</summary>

Work in three steps: (1) read the inputs, (2) calculate food, special cost, subtotal, GST = subtotal × 5 / 100, grand total, (3) print with f-strings and `:.2f`.

## 1. Indentation: Python's way of grouping code

## 2. The `if` statement
Remember the **diamond** in a flowchart from Week 2? `if` is that diamond in code. It asks a question (a condition). If the answer is `True`, the indented lines run; if it is `False`, they are skipped.

**Syntax:** `if condition:` followed by indented lines.

````python
marks = 72
if marks >= 50:              # the condition: is 72 >= 50? -> True
    print("Pass")            # runs only when the condition is True
print("Program finished")    # not indented, so it always runs
````

> **What happened?** The condition was True, so `Pass` was printed. Change `marks` to 40 and run again: `Pass` disappears, but `Program finished` is still printed because it is outside the `if`.

### `if ... else`: two paths
`else` gives a second path that runs when the condition is **False**. Exactly **one** of the two blocks always runs.

````python
attendance = 68
if attendance >= 75:                          # is 68 >= 75? -> False
    print("Eligible for the exam")            # skipped
else:                                         # runs because the condition was False
    print("Not eligible - attendance shortage")
````

> **What happened?** 68 is less than 75, so the `if` block was skipped and the `else` block ran. Note that `else` has no condition of its own and ends with a colon.

### `if ... elif ... else`: many paths
`elif` means **else if**. Use it when there are more than two possible results, like grades.
Python checks the conditions **from top to bottom** and runs only the **first** one that is True. The rest are skipped. `else` at the end catches everything that did not match.

````python
marks = 83

if marks >= 90:            # 83 >= 90? False -> check next
    grade = "O"
elif marks >= 80:          # 83 >= 80? True  -> this block runs, all others are skipped
    grade = "A+"
elif marks >= 70:
    grade = "A"
elif marks >= 60:
    grade = "B+"
elif marks >= 50:
    grade = "B"
else:                      # none of the above were True
    grade = "RA (re-appear)"

print("Grade:", grade)
````

> **What happened?** 83 is also >= 70, >= 60 and >= 50, but those lines were never checked, because Python stops at the **first** True condition. That is why the order matters: always check from the highest to the lowest.
>

## 4. The `while` loop
A loop **repeats** code. A `while` loop keeps repeating **as long as** its condition is True. Each repetition is called an **iteration**.

**Important:** something inside the loop must change, so that the condition eventually becomes False. Otherwise the loop never stops (an *infinite loop*).

````python
count = 5                       # start value
while count > 0:                # repeat while count is more than 0
    print("Launch in", count)
    count -= 1                  # reduce count by 1 each time - this makes the loop end
print("Lift off!")              # runs once, after the loop finishes
````

> **What happened?** The loop ran 5 times with count = 5, 4, 3, 2, 1. When count became 0, the condition `count > 0` was False, so the loop stopped and `Lift off!` was printed.
>

````python
correct_pin = "2468"
attempts = 0                         # how many times the user has tried

while attempts < 3:                  # allow at most 3 tries
    pin = input("Enter your UPI PIN: ")
    attempts += 1                    # count this attempt
    if pin == correct_pin:
        print("Payment successful")
        break                        # correct PIN - leave the loop now
    else:
        print("Wrong PIN. Attempts left:", 3 - attempts)

if attempts == 3 and pin != correct_pin:     # used all 3 tries and still wrong
    print("Account locked for 24 hours")
````

## 5. The `for` loop and `range()`
A `for` loop repeats once for **each item** in a sequence. It is the most common loop in Python.

`range()` creates a sequence of numbers:

| Code | Numbers produced |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 (starts at 0 and stops **before** 5) |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(2, 11, 2)` | 2, 4, 6, 8, 10 (the third number is the step) |
| `range(10, 0, -1)` | 10, 9, ..., 1 (counting down) |

In `for i in range(5):`, the variable `i` takes each value in turn.

````python
for i in range(5):          # i becomes 0, then 1, 2, 3, 4
    print("Hello", i)       # this indented line runs once for each value of i
````

> **What happened?** The loop ran **5 times** and `i` went from 0 to 4 - not 1 to 5. `range(5)` starts at 0 and stops before 5.

Now a useful example: a **multiplication table**. We use `range(1, 11)` to get the numbers 1 to 10.

````python
n = 7
for i in range(1, 11):               # i = 1, 2, ..., 10
    print(f"{n} x {i:2} = {n * i}")  # {i:2} reserves 2 spaces so the columns line up
````

> **What happened?** One `print()` line inside the loop produced 10 lines of output. That is the power of loops: write once, repeat many times.

The next cell shows `range()` with a **step**: counting in twos, and counting backwards. `end=" "` keeps the numbers on one line.

````python
for even in range(2, 11, 2):      # start 2, stop before 11, step 2
    print(even, end=" ")
print()                           # an empty print() just moves to a new line

for x in range(10, 0, -1):        # start 10, stop before 0, step -1 (count down)
    print(x, end=" ")
````

> **What happened?** The first loop printed the even numbers 2 to 10. The second counted down from 10 to 1 - it stopped **before** 0.

### Looping over a string
A `for` loop can go through **each character** of a string, one at a time.

````python
word = "PYTHON"
for letter in word:     # letter becomes "P", then "Y", then "T", ...
    print(letter)
````

> **What happened?** The loop ran 6 times, once per letter. In Notebook 3 you will loop over lists and dictionaries in exactly the same way.

## 6. `break` and `continue`
Two keywords let you control a loop from inside:
- `break` - **stop** the whole loop immediately and jump to the code after it.
- `continue` - **skip** the rest of this round and go straight to the next round.

First, `break`: find the first number after 20 that is divisible by 7.

````python
for n in range(20, 40):        # check 20, 21, 22, ...
    if n % 7 == 0:             # divisible by 7?
        print("First multiple of 7 after 20 is", n)
        break                  # found it - no need to check the remaining numbers
````

> **What happened?** The loop checked 20 and 21... and found 21. `break` stopped the loop at once, so 28 and 35 were never printed.

Now `continue`: print study hours, but skip the days when the student did not study (marked 0).

````python
hours = [6, 0, 7, 5, 0, 8]     # a list of study hours for 6 days
for h in hours:
    if h == 0:
        continue               # skip the print below and move to the next day
    print("Studied", h, "hours")
````

> **What happened?** The two zero days were skipped, but the loop **kept going** - that is the difference from `break`.
> 
> `pass` is a third keyword. It means *do nothing* and is used as a placeholder when Python needs a line but you have not written the code yet.

## 7. Common loop patterns
These three patterns appear in almost every program. Learn them well!
1. **Accumulator** - start a total at 0 and add to it in each round.
2. **Counter** - start a count at 0 and add 1 when a condition is True.
3. **Find the largest** - remember the best value seen so far and replace it when you find a bigger one.

*(A list `[ ... ]` holds many values. You will learn lists in detail in Notebook 3.)*

````python
marks = [78, 45, 92, 66, 38, 85]

# Pattern 1: accumulator (running total)
total = 0                     # start with nothing
for m in marks:
    total += m                # add each mark to the total
print("Total:", total)

# Pattern 2: counter
passed = 0                    # start counting from 0
for m in marks:
    if m >= 50:
        passed += 1           # count only the marks that pass
print("Students passed:", passed)

# Pattern 3: find the largest
highest = marks[0]            # assume the first mark is the highest
for m in marks:
    if m > highest:           # found a bigger one?
        highest = m           # remember it
print("Highest:", highest)
````

> **What happened?** Total = 404, 4 students passed (78, 92, 66, 85), and the highest mark is 92. Python also has the built-in functions `sum()` and `max()` for this, but you must understand these patterns, because many problems need a version that no built-in function covers.

### AI connection: calculating accuracy with a loop
In Week 3 you counted correct predictions by hand. Here a loop compares the **actual** labels with the model's **predicted** labels, one by one, and counts the matches. `len()` gives the number of items in a list.

````python
actual    = ["dry", "wet", "dry", "wet", "dry", "wet", "dry", "wet", "dry", "wet"]
predicted = ["dry", "wet", "wet", "wet", "dry", "wet", "dry", "dry", "dry", "wet"]

correct = 0
for i in range(len(actual)):           # i = 0, 1, ..., 9 (one index per test image)
    if actual[i] == predicted[i]:      # compare the two labels at position i
        correct += 1                   # counter pattern

print(f"Correct: {correct} / {len(actual)}")
print(f"Accuracy: {correct / len(actual) * 100:.0f}%")
````

## 8. Nested loops
A **nested loop** is a loop inside another loop. The inner loop runs **completely** for **each** round of the outer loop. It is useful for patterns, tables and grids.

````python
rows = 5
for i in range(1, rows + 1):      # outer loop: one round per row (i = 1 to 5)
    for j in range(i):            # inner loop: print i stars in this row
        print("*", end=" ")
    print()                       # after each row, move to a new line
````

> **What happened?** In row 1 the inner loop ran once, in row 2 twice, and so on, making a triangle. The empty `print()` belongs to the **outer** loop (look at its indentation), so it runs once per row.

Python can also **repeat a string** with `*`: `"* " * 3` gives `"* * * "`. This makes the same pattern with only one loop.

````python
for i in range(1, 6):
    print("* " * i)       # repeat "* " i times
````

> **What happened?** Same triangle, shorter code. There is often more than one correct way to solve a problem in Python.

Use square brackets to get one character: `word[0]`.

````python
word = "PYTHON"
print(word[0])      # index 0 -> first character 'P'
print(word[3])      # index 3 -> fourth character 'H'
print(word[-1])     # index -1 -> last character 'N'
print(len(word))    # len() gives the number of characters -> 6
````

> **What happened?** `word[3]` is **H**, the fourth letter, because counting starts at 0. The last valid index is `len(word) - 1` = 5. Asking for `word[6]` would give an `IndexError`.

### Slicing: `text[start:stop]`
A **slice** takes a part of a string. It starts at `start` and stops **before** `stop` (just like `range`).
- Leave out `start` to begin at the start; leave out `stop` to go to the end.
- `[::-1]` uses a step of -1, which **reverses** the string.

````python
dept = "Artificial Intelligence"
print(dept[0:10])     # characters 0 to 9 -> 'Artificial'
print(dept[11:])      # from index 11 to the end -> 'Intelligence'
print(dept[:3])       # from the start up to index 2 -> 'Art'
print(dept[::-1])     # step -1 -> the whole string reversed
````

> **What happened?** `dept[0:10]` gave the 10 characters at positions 0-9. Index 10 is the space, so `Intelligence` starts at index 11. The reversed string is useful for checking palindromes (see the exercises).

### Useful string methods
A **method** is a function that belongs to a value. Call it with a dot: `text.upper()`.
String methods **return a new string** - they do not change the original.

| Method | What it does |
|---|---|
| `strip()` | removes spaces at both ends |
| `upper()` / `lower()` / `title()` | CAPITALS / small letters / Each Word Capitalised |
| `replace(a, b)` | replaces `a` with `b` |
| `find(x)` | position of `x` (-1 if not found) |
| `count(x)` | how many times `x` appears |
| `startswith(x)` / `endswith(x)` | True/False |

````python
name = "  asha kumar  "
print(name.strip())            # 'asha kumar' - spaces at both ends removed
print(name.strip().title())    # methods can be chained: strip, then title -> 'Asha Kumar'
print(name.upper())
print(name.lower())

email = "asha.k@bitsathy.ac.in"
print(email.endswith(".ac.in"))               # does it end with .ac.in? -> True
print(email.find("@"))                        # position of '@' -> 6
print(email.count("a"))                       # how many 'a' characters
print(email.replace("asha.k", "ravi.m"))      # a new string with the name replaced
````

> **What happened?** Chaining `.strip().title()` first removed the spaces and then capitalised each word. `find("@")` returned 6 - the index where `@` is. Cleaning text like this is the first step in every language-AI (NLP) project.

### `split()` and `join()`
- `split()` breaks a string into a **list** of pieces. With no argument it splits at spaces; `split(",")` splits at commas.
- `join()` does the opposite: it glues a list of strings into one string, with the chosen text between the pieces.

````python
sentence = "AI learns patterns from data"
words = sentence.split()             # split at spaces -> a list of 5 words
print(words)
print("Number of words:", len(words))

print("-".join(words))               # glue the words together with '-' between them

marks_line = "78,85,92,66"
print(marks_line.split(","))         # split at commas -> ['78', '85', '92', '66']
````

> **What happened?** `split()` turned one sentence into a list of words, which is how a computer starts to 'read' text. Note that `"78,85,92,66".split(",")` gives **strings** like `'78'` - convert them with `int()` before doing maths.

### Strings cannot be changed in place
Strings are **immutable** (unchangeable). `word[0] = "J"` gives a `TypeError`. To 'change' a string, build a **new** one from pieces.

````python
word = "Python"
new_word = "J" + word[1:]     # 'J' + 'ython' -> a brand-new string
print(new_word)
print(word)                   # the original is unchanged
````

> **What happened?** We took the slice `word[1:]` (everything except the first letter) and joined `J` in front. The original `word` is still `Python`.

## 2. Lists: many values in one variable
Imagine storing marks of 60 students in 60 separate variables - impossible! A **list** stores many values in **one** variable.
- Write a list with square brackets `[ ]` and commas between items.
- Items keep their **order** and have indexes (from 0), just like strings.
- Lists are **mutable**: you can add, remove and change items.

````python
marks = [78, 85, 92, 66, 70]
subjects = ["Maths", "Physics", "Chemistry", "Python", "English"]

print(marks)
print(subjects[0])        # first item
print(subjects[-1])       # last item
print(marks[1:3])         # slice: items at index 1 and 2 -> [85, 92]
print(len(marks))         # number of items -> 5
````

> **What happened?** Indexing and slicing work exactly as with strings. A slice of a list gives another list.

### Changing a list
| Method | What it does |
|---|---|
| `append(x)` | add `x` at the end |
| `insert(i, x)` | add `x` at position `i` |
| `list[i] = x` | replace the item at position `i` |
| `remove(x)` | remove the first `x` found |
| `pop()` | remove the last item and give it back |

````python
fruits = ["apple", "banana"]
fruits.append("mango")          # add at the end
fruits.insert(1, "orange")      # add at position 1; others move right
print(fruits)

fruits[0] = "green apple"       # replace the item at index 0
print(fruits)

fruits.remove("banana")         # remove by value
last = fruits.pop()             # remove the last item and store it in 'last'
print("Removed:", last)
print(fruits)
````

> **What happened?** Unlike strings, the **same** list changed after each step. Watch how `insert(1, ...)` pushed `banana` from index 1 to index 2.

### Built-in functions and sorting
Python has functions that work on a whole list at once: `sum()`, `max()`, `min()`, `len()`.
- `list.sort()` sorts the list **itself** (changes it).
- `sorted(list)` gives a **new** sorted list and leaves the original unchanged.
- `x in list` checks whether `x` is in the list (True/False).

````python
marks = [78, 85, 92, 66, 70]
print("Total:", sum(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print(f"Average: {sum(marks) / len(marks):.2f}")   # total divided by count

marks.sort()                       # sort the list itself, smallest first
print("Ascending:", marks)
marks.sort(reverse=True)           # largest first
print("Descending:", marks)

print(sorted(["Ravi", "Asha", "Kiran"]))   # a new list in alphabetical order
print(85 in marks)                         # is 85 in the list? -> True
````

> **What happened?** One line each gave the total, highest, lowest and average - compare this with the loop patterns of Notebook 2. Average = `sum / len` is the most used formula in data analysis.

### Looping over lists
- `for item in list:` - visit each item.
- `enumerate()` - gives the **position and the item** together (we start counting at 1 here).
- `zip()` - walks through **two lists side by side**.

````python
subjects = ["Maths", "Physics", "Python"]
marks = [88, 74, 95]

for s in subjects:                          # s takes each subject in turn
    print("Subject:", s)

for i, s in enumerate(subjects, start=1):   # i = 1, 2, 3 and s = each subject
    print(i, s)

for s, m in zip(subjects, marks):           # pairs: (Maths, 88), (Physics, 74), (Python, 95)
    print(f"{s:10} {m}")                    # {s:10} uses 10 spaces so the columns line up
````

> **What happened?** `zip()` matched each subject with its mark by position. This is how you compare actual and predicted labels in AI - item 1 with item 1, item 2 with item 2, and so on.

### Building a list with a loop
Start with an **empty list** `[]` and `append()` a new item in each round of the loop.

````python
squares = []                    # empty list
for n in range(1, 6):
    squares.append(n * n)       # add the square of n to the list
print(squares)
````

> **What happened?** The list grew from `[]` to `[1, 4, 9, 16, 25]`, one item per round.

### List comprehension (a short way to build lists)
Python lets you build a list in **one line**:

`[expression for item in sequence if condition]`

Read it as: *"make a list of `expression`, for each `item` in `sequence`, but only if `condition` is True"*. The `if` part is optional.

````python
squares = [n * n for n in range(1, 6)]            # same result as the loop above
print(squares)

marks = [78, 45, 92, 38, 85]
passed = [m for m in marks if m >= 50]            # keep only the marks >= 50
print("Passed marks:", passed)

names = ["asha", "ravi", "kiran"]
print([n.title() for n in names])                 # change every item
````

> **What happened?** Each comprehension replaced a 3-4 line loop. Use them for simple cases; for complicated logic a normal loop is easier to read.

## 3. Tuples: lists that cannot change
A **tuple** is like a list but written with round brackets `( )` and it **cannot be changed** after it is created.
Use it for values that belong together and should stay fixed, such as a date or a location.

**Unpacking** puts each item of a tuple into its own variable in one line.

````python
campus_location = (11.4966, 77.2770)   # (latitude, longitude) - approximate
dob = (15, 8, 2008)                    # (day, month, year)

print(campus_location[0])              # index works like a list
day, month, year = dob                 # unpacking: 3 items into 3 variables
print(f"Born on {day}/{month}/{year}")
````

> **What happened?** Reading items works the same as a list, but `dob[0] = 20` would give an error. That protection is the reason to use a tuple for fixed data.

## 4. Dictionaries: data with labels
A list finds items by **position** (0, 1, 2...). A **dictionary** finds items by a **name** called a **key**.
It stores **key: value** pairs inside curly brackets `{ }`, just like a real dictionary stores *word: meaning*.

Use `dict[key]` to get a value. Use `dict.get(key, default)` when the key might be missing - it returns the default instead of an error.

````python
student = {
    "name": "Asha",       # key "name"  -> value "Asha"
    "roll": 101,
    "dept": "AI and DS",
    "cgpa": 8.6,
}

print(student["name"])                      # look up by key
print(student["cgpa"])
print(student.get("phone", "not given"))    # key missing -> returns "not given"
````

> **What happened?** `student["name"]` is much clearer than `student[0]` - you do not have to remember positions. `student["phone"]` would give a `KeyError`, but `.get()` returned the default safely.

### Adding, changing and removing
- `dict[key] = value` **adds** a new pair if the key is new, or **changes** the value if the key already exists.
- `del dict[key]` removes a pair.
- `key in dict` checks whether a key exists.

````python
student = {"name": "Asha", "roll": 101}
student["hostel"] = True        # new key -> added
student["roll"] = 102           # existing key -> value changed
print(student)

del student["hostel"]           # remove the pair with key "hostel"
print(student)
print("name" in student)        # does the key "name" exist? -> True
````

> **What happened?** The same syntax `student[key] = value` both added `hostel` and updated `roll`. Keys are unique - a dictionary cannot have two `roll` keys.

### Looping over a dictionary
- `for key in dict:` gives each **key**.
- `for key, value in dict.items():` gives each **key and value** together - the most useful form.
- `.keys()` and `.values()` give all keys or all values.

````python
marks = {"Maths": 88, "Physics": 74, "Python": 95}

for subject in marks:                    # loops over the keys
    print(subject)

for subject, mark in marks.items():      # key and value together
    print(f"{subject}: {mark}")

print("Keys:", list(marks.keys()))
print("Values:", list(marks.values()))
print("Total:", sum(marks.values()))     # add up all the values
````

> **What happened?** `.items()` gave each subject with its mark. `sum(marks.values())` added only the numbers, giving 257.

### AI connection: counting words
Spam filters and chatbots begin by **counting words**. A dictionary is perfect for this: the **word** is the key and its **count** is the value.
For each word: if it is already in the dictionary, add 1; otherwise, start its count at 1.

````python
message = "win win free money click now win free gift"
counts = {}                          # empty dictionary
for word in message.split():         # go through each word
    if word in counts:
        counts[word] += 1            # seen before -> increase its count
    else:
        counts[word] = 1             # first time -> start at 1
print(counts)
print("Most common word:", max(counts, key=counts.get))   # key with the biggest value
````

> **What happened?** `win` appears 3 times and `free` twice. Turning text into word counts like this is called a **bag of words** - one of the oldest ways to feed text into a machine learning model. (`max(counts, key=counts.get)` finds the key whose value is largest.)

## 5. Sets: unique items only
A **set** is written with `{ }` (without key: value pairs). It has two special properties:
- **No duplicates** - each item appears only once.
- **No order** - items have no index, and the printed order may vary.

Sets are great for removing repeats and for comparing groups:
`&` items in **both**, `|` items in **either**, `-` items in the first but **not** the second.

````python
visitors = ["Asha", "Ravi", "Asha", "Kiran", "Ravi"]
unique = set(visitors)                 # duplicates removed automatically
print(unique)
print("Unique visitors:", len(unique))

asha_courses = {"Python", "AI", "Maths", "English"}
ravi_courses = {"Python", "Physics", "Maths"}

print("Both:", asha_courses & ravi_courses)       # intersection - in both sets
print("Either:", asha_courses | ravi_courses)     # union - in at least one set
print("Only Asha:", asha_courses - ravi_courses)  # difference - in Asha's but not Ravi's
````

> **What happened?** Five names became 3 unique visitors. The set operations answered *who takes which course* in one line each. Do not worry if the order of items is different on your screen - sets have no order.

## 7. A small dataset: list of dictionaries
In Week 3 you stored class data in a spreadsheet: **each row is a record, each column is a feature**.
In Python, the same table is a **list of dictionaries**: each dictionary is one **row**, and its keys are the **column names**.
The cell below stores 5 rows and prints them as a neat table.

````python
students = [
    {"roll": 101, "name": "Asha",  "type": "Hostel",      "sleep": 7.0, "marks": 86},
    {"roll": 102, "name": "Ravi",  "type": "Day scholar", "sleep": 6.0, "marks": 72},
    {"roll": 103, "name": "Kiran", "type": "Day scholar", "sleep": 6.5, "marks": 64},
    {"roll": 104, "name": "Meena", "type": "Hostel",      "sleep": 8.0, "marks": 91},
    {"roll": 105, "name": "Arun",  "type": "Hostel",      "sleep": 5.5, "marks": 58},
]

# Header row. :<6 means 'left-align in 6 spaces', :>6 means 'right-align in 6 spaces'
print(f"{'Roll':<6}{'Name':<8}{'Type':<13}{'Sleep':>6}{'Marks':>7}")
for s in students:                     # s is one row (one dictionary)
    print(f"{s['roll']:<6}{s['name']:<8}{s['type']:<13}{s['sleep']:>6}{s['marks']:>7}")
````

> **What happened?** Each loop round took one dictionary (row) and printed its values in fixed-width columns. Inside an f-string we used single quotes `s['name']`, because the f-string itself uses double quotes.

Now we can **ask questions** about the data with the tools from this notebook: a comprehension to filter rows, `sum()/len()` for an average, and the *find the largest* pattern for the topper.

````python
# 1. Filter: keep only the rows where type is "Hostel"
hostel = [s for s in students if s["type"] == "Hostel"]
print("Hostel students:", [s["name"] for s in hostel])

# 2. Average of one column for the filtered rows
avg_sleep = sum(s["sleep"] for s in hostel) / len(hostel)
print(f"Average sleep of hostel students: {avg_sleep:.1f} hours")

# 3. Find the row with the highest marks
top = students[0]                    # assume the first student is the topper
for s in students:
    if s["marks"] > top["marks"]:    # found a higher mark?
        top = s
print("Topper:", top["name"], top["marks"])
````

> **What happened?** Three hostel students sleep 6.8 hours on average, and Meena is the topper. Next semester you will use **Pandas**, which answers questions like these on thousands of rows in one line. Now you know what happens underneath!

## Mini project: Class result analyser
**Mini project.** You have a small class dataset:
```python
results = {"Asha": [88, 92, 79], "Ravi": [45, 60, 52], "Kiran": [35, 48, 41], "Meena": [95, 89, 97], "Arun": [70, 66, 58]}
```
For each student print the **total**, **average** and **result** (`Pass` if every subject is 50 or more, else `Fail`). Then print the **class topper** and the **pass percentage** of the class.

<details><summary><b>Hint</b> (click to open)</summary>

Loop with `for name, marks in results.items():`. A student passes if `min(marks) >= 50`. Use a counter for passes and the find-the-largest pattern for the topper.

</details>

````python
results = {"Asha": [88, 92, 79], "Ravi": [45, 60, 52], "Kiran": [35, 48, 41], "Meena": [95, 89, 97], "Arun": [70, 66, 58]}

topper = ""
highest_total = -1
pass_count = 0
total_students = len(results)

for student, marks in results.items():
    total = sum(marks)
    average = total / len(marks)
    
    # Check if every subject is 50 or more
    if all(m >= 50 for m in marks):
        status = "Pass"
        pass_count += 1
    else:
        status = "Fail"
        
    print(f"Student: {student} | Total: {total} | Average: {average:.2f} | Result: {status}")
    
    if total > highest_total:
        highest_total = total
        topper = student

pass_percentage = (pass_count / total_students) * 100

print(f"\nClass Topper: {topper} with a total of {highest_total}")
print(f"Pass Percentage: {pass_percentage:.2f}%")
````

## 1. Why functions?
A **function** is a named block of code that does one job. You **define** it once and then **call** it (use it) as many times as you like.
You have already used built-in functions: `print()`, `len()`, `input()`, `sum()`, `max()`. Now you will write your own.

**Why bother?**
- No repeated code - write once, use many times.
- Easier to read - a good name like `calculate_gpa()` explains what the code does.
- Easier to fix - a bug is fixed in one place only.

**Syntax:**
```
def function_name():      # def = define; ends with a colon
    indented lines        # the body of the function
```

````python
def greet():                                          # define a function called greet
    print("Welcome to the Python lab!")               # the body (indented)
    print("Please log in with your roll number.")

greet()      # call the function - its body runs now
greet()      # call it again - the body runs again
````

> **What happened?** Defining the function printed **nothing** - `def` only stores the instructions. The messages appeared only when we **called** `greet()`. Each call ran the whole body, so we got the two lines twice.

### Parameters: giving input to a function
A function becomes more useful when it can work with different data. Names inside the brackets of `def` are **parameters** - placeholders for values.
The actual values you pass when calling are called **arguments**.

````python
def greet(name, dept):                       # name and dept are parameters
    print(f"Hello {name} from {dept}!")

greet("Asha", "AI and DS")                   # arguments: name = "Asha", dept = "AI and DS"
greet("Ravi", "Mechanical")                  # same function, different data
````

> **What happened?** In the first call, `name` held "Asha"; in the second, "Ravi". Arguments are matched to parameters **by position**: first to first, second to second.

### `return`: giving a result back
`print()` only **shows** a value on the screen. `return` **sends the value back** to the place where the function was called, so you can store it in a variable, use it in a calculation, or check it with `if`.
When Python reaches `return`, the function ends immediately.

````python
def average(a, b, c):
    return (a + b + c) / 3          # send the result back to the caller

avg = average(80, 90, 70)           # the returned value (80.0) is stored in avg
print("Average:", avg)

if average(40, 45, 50) < 50:        # the returned value can be used directly in a condition
    print("Needs improvement")
````

> **What happened?** `average(80, 90, 70)` was replaced by its result, 80.0, which we stored in `avg`. In the `if`, the result 45.0 was compared with 50 straight away.

### `print` vs `return` - the most important difference
Run the cell and compare the values of `x` and `y`.
A function with no `return` gives back the special value `None` (*nothing*).

````python
def add_print(a, b):
    print(a + b)          # only SHOWS the answer

def add_return(a, b):
    return a + b          # GIVES BACK the answer

x = add_print(2, 3)       # shows 5 on screen, but x receives nothing (None)
y = add_return(2, 3)      # shows nothing, but y receives 5
print("x =", x, " y =", y)
````

> **What happened?** `x` is `None` because `add_print` never returned anything - it only printed. `y` is 5. **Rule of thumb:** if you need to use the answer later, `return` it.

### Functions that make decisions
A function can contain `if/elif/else`. Each branch can `return` a different value. Once a `return` runs, the function stops.

````python
def grade(marks):
    if marks >= 90:
        return "O"          # function ends here for marks >= 90
    elif marks >= 80:
        return "A+"
    elif marks >= 70:
        return "A"
    elif marks >= 60:
        return "B+"
    elif marks >= 50:
        return "B"
    else:
        return "RA"

for m in [95, 83, 71, 49]:      # call the same function for four marks
    print(m, "->", grade(m))
````

> **What happened?** The grade logic from Notebook 2 is now packed into a reusable function. Any program can call `grade()` without knowing how it works inside.

### Default values and keyword arguments
- A **default value** (`cost_per_day=120`) is used when the caller does not give that argument.
- A **keyword argument** (`gst=0`) names the parameter when calling, so the order does not matter.

````python
def mess_bill(days, cost_per_day=120, gst=5):     # two parameters have defaults
    subtotal = days * cost_per_day
    return subtotal + subtotal * gst / 100

print(mess_bill(30))                  # uses cost_per_day=120 and gst=5
print(mess_bill(30, 150))             # cost_per_day=150, gst still 5
print(mess_bill(days=25, gst=0))      # keyword arguments: cost stays 120, gst becomes 0
````

> **What happened?** Call 1: 30 × 120 + 5% = 3780.0. Call 2: 30 × 150 + 5% = 4725.0. Call 3: 25 × 120 with no GST = 3000.0. Defaults make functions easy to use for the common case but still flexible.

### Returning more than one value
A function can return several values separated by commas. The caller **unpacks** them into variables (like tuple unpacking in Notebook 3).

````python
def stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)   # three values

low, high, avg = stats([56, 78, 91, 45, 67])     # unpack into three variables
print(f"Lowest {low}, highest {high}, average {avg:.1f}")
````

> **What happened?** Python packed the three values into a tuple and the caller unpacked them in order: `low` = 45, `high` = 91, `avg` = 67.4.

### Docstrings
A **docstring** is a string in triple quotes written just below `def`. It explains what the function does. `help(function_name)` shows it.
Good programmers write one for every function.

````python
def accuracy(correct, total):
    """Return the accuracy as a percentage, e.g. accuracy(9, 10) -> 90.0"""
    return correct / total * 100

print(accuracy(9, 10))
help(accuracy)          # displays the docstring
````

> **What happened?** `help()` printed the function's name, parameters and docstring. In Colab you can also hover over a function name to see its docstring.

### Local and global variables
- A variable created **outside** all functions is **global** - it can be read everywhere.
- A variable created **inside** a function is **local** - it exists only while that function runs and cannot be used outside.

This keeps functions independent: a variable named `total` inside one function does not disturb a `total` elsewhere.

````python
college = "BIT"          # global variable

def show():
    year = 1             # local variable - lives only inside show()
    print(college, "year", year)   # a function can read a global variable

show()
print(college)           # works: college is global
# print(year)            # remove the # and run: NameError, because year is local to show()
````

> **What happened?** `show()` could use both `college` and `year`. Outside the function only `college` exists. Try removing the `#` on the last line to see the `NameError` for yourself.

### Extra: `lambda` (small one-line functions)
A `lambda` is a tiny function without a name, written in one line: `lambda parameters: expression`.
You will often see it in AI code, mostly to tell `sort()` or `max()` **what to compare**.

````python
square = lambda x: x * x                  # same as: def square(x): return x * x
print(square(6))

students = [("Asha", 86), ("Ravi", 72), ("Meena", 91)]
students.sort(key=lambda s: s[1], reverse=True)    # sort by item [1] (the marks), largest first
print(students)
````

> **What happened?** `key=lambda s: s[1]` told `sort()` to compare the students by their marks (index 1 of each tuple) instead of by their names. `lambda` is optional for this course - a normal `def` function always works too.

## 2. Modules: ready-made code
A **module** is a file of useful functions written by other people. Python comes with hundreds of them. Use `import` to bring a module in, then use its functions with a dot: `module.function()`.

| Module | Useful for |
|---|---|
| `math` | square root, pi, rounding up/down, factorial |
| `random` | random numbers, random choices, shuffling |
| `datetime` | today's date, differences between dates |

First, the `math` module:

````python
import math                           # load the math module

print(math.sqrt(144))                 # square root -> 12.0
print(math.pi)                        # the value of pi
print(math.ceil(4.1), math.floor(4.9))  # round UP -> 5, round DOWN -> 4
print(math.factorial(5))              # 5! -> 120
````

> **What happened?** We did not write any of these functions - we just used them. `math.factorial(5)` gives the same 120 that you calculated with a loop in Notebook 2.

The `random` module makes programs unpredictable - useful for games, quizzes and for **shuffling data** before training an AI model. Run the cell several times: the output changes each time.

````python
import random

print(random.randint(1, 6))                          # random whole number from 1 to 6 (a dice)
print(random.choice(["rock", "paper", "scissors"]))  # pick one item from a list

names = ["Asha", "Ravi", "Kiran", "Meena"]
random.shuffle(names)                                # mix the list into a random order
print(names)
````

> **What happened?** Every run gives different results. In machine learning, data is shuffled before splitting it into training and test sets, so that the split is fair.

The `datetime` module works with dates. Subtracting two dates gives the number of days between them. Here we use `from ... import ...` to bring in only `date`.

````python
from datetime import date            # import only the 'date' part of the module

today = date.today()                 # today's date from the computer's clock
print("Today:", today)
exam = date(2027, 4, 20)             # a sample exam date: year, month, day
print("Days left for the exam:", (exam - today).days)   # difference in days
````

> **What happened?** `exam - today` gives a time difference, and `.days` turns it into a whole number of days. Your answer depends on the day you run it.

## 3. Working with files
Data in variables **disappears** when the program ends. **Files** keep data saved for later.

To use a file: **open** it, **read or write**, then **close** it. `with open(...) as f:` does all of this and closes the file automatically, even if an error happens.

| Mode | Meaning |
|---|---|
| `"w"` | **write** - creates a new file; **erases** old content if the file exists |
| `"a"` | **append** - adds to the end of the file |
| `"r"` | **read** - the default |

`"\n"` is the **new line** character. `write()` does not add it automatically.

*Colab note: files are saved in the Colab session (click the folder icon on the left to see them). They are deleted when the session ends - download them if you need them.*

````python
with open("marks.txt", "w") as f:    # open marks.txt for writing; f is our name for the file
    f.write("Asha,86\n")             # \n moves to a new line in the file
    f.write("Ravi,72\n")
    f.write("Meena,91\n")
print("File saved")                  # this line is outside 'with' - the file is closed now
````

> **What happened?** A file called `marks.txt` now exists with 3 lines. Open the folder icon on the left of Colab to see it (you may need to click refresh).

Now we **append** one more line (mode `"a"` keeps the old content), and then **read** the whole file with `read()`.

````python
with open("marks.txt", "a") as f:    # append mode - add to the end, keep old lines
    f.write("Kiran,64\n")

with open("marks.txt", "r") as f:    # read mode
    content = f.read()               # read the whole file into one string
print(content)
````

> **What happened?** All 4 lines were printed: the 3 old ones plus Kiran. If we had used `"w"` instead of `"a"`, the file would contain only Kiran.

### Reading line by line and using the data
A `for` loop over a file gives **one line at a time**. For each line we:
1. `strip()` the `\n` at the end,
2. `split(",")` it into name and mark,
3. convert the mark to `int` so we can calculate.

````python
total = 0
count = 0
with open("marks.txt", "r") as f:
    for line in f:                              # one line at a time, e.g. "Asha,86\n"
        name, mark = line.strip().split(",")    # -> "Asha" and "86"
        mark = int(mark)                        # text -> number
        total += mark                           # accumulator
        count += 1                              # counter
        print(f"{name:<6} {mark}")
print(f"Class average: {total / count:.2f}")
````

> **What happened?** The program read the data back from the file and calculated the class average, 78.25. A file whose values are separated by commas is called a **CSV** (comma-separated values) file. Google Sheets can save data as CSV, and most AI datasets are CSV files.

## 4. Handling errors with `try` and `except`
When something goes wrong (the user types letters instead of a number, a file is missing...), Python stops with an error - the program **crashes**.
`try ... except` lets the program **catch** the error and handle it politely:
```
try:
    code that might fail
except ErrorType:
    what to do if that error happens
```
Run the cell below **twice**: first type `18`, then type `eighteen`.

````python
try:
    age = int(input("Enter your age: "))     # int() fails if the text is not a number
    print("Next year you will be", age + 1)
except ValueError:                           # runs only if a ValueError happened above
    print("Please enter a number, like 18")
````

> **What happened?** With `18`, the `try` block finished normally and the `except` block was skipped. With `eighteen`, `int()` raised a `ValueError`; Python jumped to `except` and printed a friendly message instead of crashing.

Different problems raise different errors. Here we catch `ZeroDivisionError` inside a function and return `None` (*no result*) instead of crashing.

````python
def safe_divide(a, b):
    try:
        return a / b                       # may fail if b is 0
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None                        # no valid result

print(safe_divide(10, 2))                  # works normally -> 5.0
print(safe_divide(10, 0))                  # handled -> message, then None
````

> **What happened?** The first call returned 5.0. The second call hit the error, printed the message and returned `None`, and the program **kept running**.

Opening a file that does not exist raises `FileNotFoundError`.
`finally` is an optional block that **always** runs, whether there was an error or not - useful for clean-up messages.

````python
try:
    with open("attendance.txt", "r") as f:   # this file was never created
        print(f.read())
except FileNotFoundError:
    print("attendance.txt does not exist yet")
finally:
    print("This line always runs")
````

> **What happened?** Because the file is missing, the `except` block ran, and then `finally` ran as well.

### Keep asking until the input is valid
Combine a `while True` loop (repeat forever) with `try/except`. The loop only ends when `return` gives back a valid number.
This function is a handy tool you can reuse in any program that reads marks.

````python
def read_marks(prompt):
    while True:                                  # keep asking...
        try:
            m = int(input(prompt))
            if 0 <= m <= 100:                    # valid range?
                return m                         # ...until we get a good value (return ends the loop)
            print("Marks must be between 0 and 100")
        except ValueError:
            print("That is not a number. Try again.")

marks = read_marks("Enter marks (0-100): ")
print("You entered", marks)
````

> **What happened?** Try typing `abc`, then `150`, then `75`. The first two are rejected with a message, and the loop asks again. Only a valid mark is returned.

## 5. AI connection: a tiny 'model' as a function
A trained AI model behaves like a function: **features go in, a prediction comes out**.
Below, a **rule-based** function predicts whether a banana is ripe from two features (how yellow it is, and how soft it is, on a 0-100 scale). Then we test it on 5 labelled examples and calculate the accuracy - just like testing your Teachable Machine model.

````python
def predict_ripe(yellow_percent, softness):
    """Predict 'ripe' or 'unripe' for a banana from two features (0-100 scale)."""
    if yellow_percent >= 70 and softness >= 40:     # our hand-written rule
        return "ripe"
    return "unripe"

# Test data: (yellow %, softness, actual label)
test_data = [(90, 60, "ripe"), (40, 20, "unripe"), (75, 30, "ripe"), (95, 80, "ripe"), (30, 50, "unripe")]

correct = 0
for yellow, soft, actual in test_data:              # unpack each tuple
    pred = predict_ripe(yellow, soft)               # the 'model' makes a prediction
    if pred == actual:
        correct += 1
        mark = "correct"
    else:
        mark = "WRONG"
    print(f"yellow={yellow:3} soft={soft:3} -> predicted {pred:7} actual {actual:7} {mark}")

print(f"Accuracy: {correct / len(test_data) * 100:.0f}%")
````

> **What happened?** The third banana (yellow 75, softness 30) is actually ripe, but our rule said unripe, so accuracy is 80%. How would you change the rule? Trying rules, measuring accuracy and adjusting is exactly what **training** does - automatically, with thousands of examples.

## 6. Putting it together: Student Result Manager
This program uses **everything** from the four notebooks: functions, a dictionary of lists, loops, `if`, f-strings, files and `try/except`.
Read the docstring and comments of each function, then run the cell.

| Function | Job |
|---|---|
| `add_student()` | add a student and their marks to the dictionary |
| `result()` | decide Pass or Fail |
| `save()` | write all students to a CSV file |
| `load()` | read the CSV file back into a dictionary (safely) |
| `report()` | print a neat result table |

````python
def add_student(db, name, marks):
    """Add a student and their list of marks to the database dictionary."""
    db[name] = marks

def result(marks):
    """Return 'Pass' if every mark is 50 or more, otherwise 'Fail'."""
    return "Pass" if min(marks) >= 50 else "Fail"

def save(db, filename):
    """Write each student as one line: name,mark1,mark2,..."""
    with open(filename, "w") as f:
        for name, marks in db.items():
            f.write(name + "," + ",".join(str(m) for m in marks) + "\n")

def load(filename):
    """Read the file back into a dictionary. Start empty if the file is missing."""
    db = {}
    try:
        with open(filename) as f:
            for line in f:
                parts = line.strip().split(",")          # ['Asha', '88', '92', '79']
                db[parts[0]] = [int(x) for x in parts[1:]]   # name -> list of int marks
    except FileNotFoundError:
        print("No saved file found - starting fresh")
    return db

def report(db):
    """Print total, average and result for every student."""
    print(f"{'Name':<8}{'Total':>6}{'Avg':>7}  Result")
    for name, marks in db.items():
        print(f"{name:<8}{sum(marks):>6}{sum(marks) / len(marks):>7.1f}  {result(marks)}")

# ---- main program: uses the functions above ----
students = {}
add_student(students, "Asha", [88, 92, 79])
add_student(students, "Ravi", [45, 60, 52])
add_student(students, "Meena", [95, 89, 97])
save(students, "results.csv")       # store on disk

loaded = load("results.csv")        # read back from disk
report(loaded)
````

> **What happened?** The data was saved to `results.csv`, read back, and reported. Because each job is in its own small function, the **main program** at the bottom reads almost like English. This is how real software - and real AI pipelines - are organised.

### Exercise 10: Accuracy function (AI from Zero link)
Write a function `accuracy(actual, predicted)` that takes two lists of labels and returns the accuracy in %. Test it with:
```python
actual    = ["rock", "paper", "scissors", "rock", "paper"]
predicted = ["rock", "paper", "rock", "rock", "scissors"]
```
*Expected:* 60.0

<details><summary><b>Hint</b> (click to open)</summary>

Use `zip(actual, predicted)` to compare item by item and count the matches.

</details>

````python
def accuracy(actual, predicted):
    correct = sum(1 for a, p in zip(actual, predicted) if a == p)
    return (correct / len(actual)) * 100

actual    = ["rock", "paper", "scissors", "rock", "paper"]
predicted = ["rock", "paper", "rock", "rock", "scissors"]

print(accuracy(actual, predicted))
````

## Final mini project: Quiz app
**Final mini project.** Build a small quiz app:
1. Store 5 questions and answers in a **dictionary** (for example AI from Zero questions).
2. Write a function `ask(question, answer)` that asks the question with `input()`, compares the answer (ignore capitals and spaces) and returns `True` or `False`.
3. Loop through all questions and count the score.
4. Save the player's name and score to `scores.txt` using append mode.
5. Print the score and a message: 5 = *Excellent*, 3-4 = *Good*, below 3 = *Keep practising*.

<details><summary><b>Hint</b> (click to open)</summary>

Store questions as keys and answers (in lower case) as values. In `ask()`, return `reply.strip().lower() == answer`. Loop with `.items()`, count the score, then write to the file with mode `"a"`.

</details>

````python

quiz = {
    "What is the capital of France?": "paris",
    "Which planet is known as the Red Planet?": "mars",
    "What is 5 + 7?": "12",
    "Which programming language is this?": "python",
    "What is the powerhouse of the cell?": "mitochondria"
}


def ask(question, answer):
    user_ans = input(question + " ").strip().lower()
    return user_ans == answer.strip().lower()

score = 0
for q, a in quiz.items():
    if ask(q, a):
        print("Correct!\n")
        score += 1
    else:
        print(f"Wrong! The correct answer was: {a}\n")


player_name = input("Enter your name: ")
with open("scores.txt", "a") as file:
    file.write(f"Player: {player_name} | Score: {score}/5\n")

print(f"\nYour Score: {score}/5")
if score == 5:
    print("Excellent!")
elif score >= 3:
    print("Good")
else:
    print("Keep practising.")
````
